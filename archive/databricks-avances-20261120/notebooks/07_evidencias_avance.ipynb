```python

spark.sql("""
SHOW TABLES IN workspace.twitch_lakehouse
""").show(truncate=False)
```

```python

bronze = "workspace.twitch_lakehouse.bronze_interactions"

print("REGISTROS BRONZE:", spark.table(bronze).count())

spark.sql(f"DESCRIBE DETAIL {bronze}") \
    .select("format", "numFiles", "sizeInBytes") \
    .show(truncate=False)
```

```python

silver = "workspace.twitch_lakehouse.silver_interactions"
rejects = "workspace.twitch_lakehouse.silver_rejects"

bronze_rows = spark.table(bronze).count()
silver_rows = spark.table(silver).count()
rejected_rows = spark.table(rejects).count()

print({
    "bronze_rows": bronze_rows,
    "silver_rows": silver_rows,
    "rejected_rows": rejected_rows,
    "validacion_conteos": bronze_rows == silver_rows + rejected_rows
})
```

```python

from pyspark.sql import functions as F

gold = spark.table(
    "workspace.twitch_lakehouse.gold_streamer_metrics"
)

print("Streamers en Gold:", gold.count())

gold.agg(
    F.sum("interactions").alias("interacciones"),
    F.sum("estimated_minutes").alias("minutos_estimados")
).show()
```

```python

display(
    gold.orderBy(F.desc("interactions")).limit(10)
)
```
